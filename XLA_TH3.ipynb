```python
#Phần 1 Bài tập 2 (coi kĩ lại)
import numpy as np
T = 50
DT = 1

while True:

    G1 = anh_muc_xam[anh_muc_xam >= T]
    G2 = anh_muc_xam[anh_muc_xam < T]

    m1 = np.mean(G1) if G1.size > 0 else 0
    m2 = np.mean(G2) if G2.size > 0 else 0

    T_moi = (m1 + m2) / 2

    if abs(T_moi - T) < DT:
        break

    T = T_moi

print("Ngưỡng tự động:", T)

anh_ket_qua = (anh_muc_xam >= T)
plt.imshow(anh_ket_qua, cmap='gray')
plt.axis('off')
plt.show()
```

```python
from skimage.filters import threshold_local

img = cv2.imread('/content/drive/MyDrive/Colab Notebooks/Bản sao của Photo00001.jpg')


anh_muc_xam = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


plt.figure(figsize=(5,5))
plt.imshow(anh_muc_xam, cmap='gray')
plt.title("Ảnh mức xám")
plt.axis('off')
plt.show()


#toàn cục

nguong = 120

anh_toan_cuc = (anh_muc_xam >= nguong)

plt.figure(figsize=(5,5))
plt.imshow(anh_toan_cuc, cmap='gray')
plt.title(f"Phân ngưỡng toàn cục (T = {nguong})")
plt.axis('off')
plt.show()

#cục bộ

nguong_cuc_bo = threshold_local(anh_muc_xam,block_size=31,offset=10)

anh_cuc_bo = (anh_muc_xam >= nguong_cuc_bo)

plt.figure(figsize=(5,5))
plt.imshow(anh_cuc_bo, cmap='gray')
plt.title("Phân ngưỡng cục bộ")
plt.axis('off')
plt.show()
```
